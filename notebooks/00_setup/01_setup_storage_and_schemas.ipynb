# 00 · Setup: external locations and schemas
One-time, idempotent setup. Creates the Unity Catalog external locations for the `raw`, `bronze`, `silver` and `gold` ADLS containers and the `bronze`, `silver` and `gold` schemas. Safe to re-run.

```python
%run ../../config/pipeline_config
```

```python
for name, url in EXTERNAL_LOCATIONS.items():
    spark.sql(f"""
        CREATE EXTERNAL LOCATION IF NOT EXISTS `{name}`
        URL '{url}'
        WITH (STORAGE CREDENTIAL `{STORAGE_CREDENTIAL}`)
    """)
    print(f"✓ external location {name} -> {url}")
```

```python
for schema, comment in [(BRONZE_SCHEMA, "Raw data as received, plus ingestion metadata"),
                        (SILVER_SCHEMA, "Typed, cleaned and deduplicated data"),
                        (GOLD_SCHEMA,   "Star schema (dimensions and facts) and business aggregates")]:
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema} COMMENT '{comment}'")
    print(f"✓ schema {schema}")

display(spark.sql(f"SHOW SCHEMAS IN {CATALOG}"))
```
