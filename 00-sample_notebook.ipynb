```python
from pyspark.sql.functions import current_timestamp, lit, input_file_name, col, expr, regexp_replace
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType, DateType

# Storage account configuration
storage_account = "adventurework"
raw_base_path = f"abfss://raw@{storage_account}.dfs.core.windows.net/"
silver_base_path = f"abfss://silver@{storage_account}.dfs.core.windows.net/"

print("="*60)
print("INGESTING RAW -> SILVER WITH METADATA COLUMNS")
print("="*60)

# ============================================
# Helper: Add metadata columns
# ============================================
def add_metadata_columns(df):
    """Add ingestion_timestamp and source_file_name metadata columns."""
    return (
        df
        .withColumn("ingestion_timestamp", current_timestamp())
        .withColumn("source_file_name", col("_metadata.file_path"))
    )

# Raw values look like "$2,024.99" and "Friday, August 25, 2017"
def parse_money(c):
    return regexp_replace(col(c), r"[$,]", "").cast("double")

def parse_long_date(c):
    return expr(f"try_to_date(regexp_replace({c}, '^[A-Za-z]+, ', ''), 'MMMM d, yyyy')")

# ============================================
# 1. Product
# ============================================
print("\n1. Processing Product...")
product_schema = StructType([
    StructField("ProductKey", IntegerType(), True),
    StructField("ProductName", StringType(), True),
    StructField("StandardCost", StringType(), True),
    StructField("Color", StringType(), True),
    StructField("Subcategory", StringType(), True),
    StructField("Category", StringType(), True),
    StructField("BackgroundColorFormat", StringType(), True),
    StructField("FontColorFormat", StringType(), True)
])
df_Product = spark.read.option("header", True).option("sep", "\t").schema(product_schema).csv(f"{raw_base_path}Product.csv")
df_Product = df_Product.withColumn("StandardCost", parse_money("StandardCost"))
df_Product = add_metadata_columns(df_Product)
df_Product.write.mode("overwrite").option("overwriteSchema", "true").format("delta").save(f"{silver_base_path}Product")
print(f"✓ Product written to Silver: {df_Product.count()} records")
display(df_Product.limit(3))

# ============================================
# 2. Region
# ============================================
print("\n2. Processing Region...")
region_schema = StructType([
    StructField("SalesTerritoryKey", IntegerType(), True),
    StructField("Region", StringType(), True),
    StructField("Country", StringType(), True),
    StructField("Group", StringType(), True)
])
df_Region = spark.read.option("header", True).option("sep", "\t").schema(region_schema).csv(f"{raw_base_path}Region.csv")
df_Region = add_metadata_columns(df_Region)
df_Region.write.mode("overwrite").format("delta").save(f"{silver_base_path}Region")
print(f"✓ Region written to Silver: {df_Region.count()} records")
display(df_Region.limit(3))

# ============================================
# 3. Reseller
# ============================================
print("\n3. Processing Reseller...")
reseller_schema = StructType([
    StructField("ResellerKey", IntegerType(), True),
    StructField("BusinessType", StringType(), True),
    StructField("ResellerName", StringType(), True),
    StructField("City", StringType(), True),
    StructField("StateProvince", StringType(), True),
    StructField("CountryRegion", StringType(), True)
])
df_Reseller = spark.read.option("header", True).option("sep", "\t").schema(reseller_schema).csv(f"{raw_base_path}Reseller.csv")
df_Reseller = add_metadata_columns(df_Reseller)
df_Reseller.write.mode("overwrite").format("delta").save(f"{silver_base_path}Reseller")
print(f"✓ Reseller written to Silver: {df_Reseller.count()} records")
display(df_Reseller.limit(3))

# ============================================
# 4. Sales
# ============================================
print("\n4. Processing Sales...")
sales_schema = StructType([
    StructField("SalesOrderNumber", StringType(), True),
    StructField("OrderDate", StringType(), True),
    StructField("ProductKey", IntegerType(), True),
    StructField("ResellerKey", IntegerType(), True),
    StructField("EmployeeKey", IntegerType(), True),
    StructField("SalesTerritoryKey", IntegerType(), True),
    StructField("Quantity", IntegerType(), True),
    StructField("UnitPrice", StringType(), True),
    StructField("Sales", StringType(), True),
    StructField("Cost", StringType(), True)
])
df_Sales = spark.read.option("header", True).option("sep", "\t").schema(sales_schema).csv(f"{raw_base_path}Sales.csv")
df_Sales = (df_Sales
    .withColumn("OrderDate", parse_long_date("OrderDate"))
    .withColumn("UnitPrice", parse_money("UnitPrice"))
    .withColumn("Sales", parse_money("Sales"))
    .withColumn("Cost", parse_money("Cost")))
df_Sales = add_metadata_columns(df_Sales)
df_Sales.write.mode("overwrite").option("overwriteSchema", "true").format("delta").save(f"{silver_base_path}Sales")
print(f"✓ Sales written to Silver: {df_Sales.count()} records")
display(df_Sales.limit(3))

# ============================================
# 5. Salesperson
# ============================================
print("\n5. Processing Salesperson...")
salesperson_schema = StructType([
    StructField("EmployeeKey", IntegerType(), True),
    StructField("EmployeeID", StringType(), True),
    StructField("SalespersonName", StringType(), True),
    StructField("Title", StringType(), True),
    StructField("UPN", StringType(), True)
])
df_Salesperson = spark.read.option("header", True).option("sep", "\t").schema(salesperson_schema).csv(f"{raw_base_path}Salesperson.csv")
df_Salesperson = add_metadata_columns(df_Salesperson)
df_Salesperson.write.mode("overwrite").format("delta").save(f"{silver_base_path}Salesperson")
print(f"✓ Salesperson written to Silver: {df_Salesperson.count()} records")
display(df_Salesperson.limit(3))

# ============================================
# 6. SalespersonRegion
# ============================================
print("\n6. Processing SalespersonRegion...")
salesperson_region_schema = StructType([
    StructField("EmployeeKey", IntegerType(), True),
    StructField("SalesTerritoryKey", IntegerType(), True)
])
df_SalespersonRegion = spark.read.option("header", True).option("sep", "\t").schema(salesperson_region_schema).csv(f"{raw_base_path}SalespersonRegion.csv")
df_SalespersonRegion = add_metadata_columns(df_SalespersonRegion)
df_SalespersonRegion.write.mode("overwrite").format("delta").save(f"{silver_base_path}SalespersonRegion")
print(f"✓ SalespersonRegion written to Silver: {df_SalespersonRegion.count()} records")
display(df_SalespersonRegion.limit(3))

# ============================================
# 7. Targets
# ============================================
print("\n7. Processing Targets...")
targets_schema = StructType([
    StructField("EmployeeID", StringType(), True),
    StructField("Target", StringType(), True),
    StructField("TargetMonth", StringType(), True)
])
df_Targets = spark.read.option("header", True).option("sep", "\t").schema(targets_schema).csv(f"{raw_base_path}Targets.csv")
df_Targets = (df_Targets
    .withColumn("Target", parse_money("Target"))
    .withColumn("TargetMonth", parse_long_date("TargetMonth")))
df_Targets = add_metadata_columns(df_Targets)
df_Targets.write.mode("overwrite").option("overwriteSchema", "true").format("delta").save(f"{silver_base_path}Targets")
print(f"✓ Targets written to Silver: {df_Targets.count()} records")
display(df_Targets.limit(3))

# ============================================
# Summary Report
# ============================================
print("\n" + "="*60)
print("SILVER LAYER INGESTION SUMMARY")
print("="*60)

from pyspark.sql import Row

summary_data = [
    ("Product", df_Product.count(), len(df_Product.columns)),
    ("Region", df_Region.count(), len(df_Region.columns)),
    ("Reseller", df_Reseller.count(), len(df_Reseller.columns)),
    ("Sales", df_Sales.count(), len(df_Sales.columns)),
    ("Salesperson", df_Salesperson.count(), len(df_Salesperson.columns)),
    ("SalespersonRegion", df_SalespersonRegion.count(), len(df_SalespersonRegion.columns)),
    ("Targets", df_Targets.count(), len(df_Targets.columns))
]

summary_df = spark.createDataFrame(
    [Row(Table=row[0], Record_Count=row[1], Total_Columns=row[2]) for row in summary_data]
)
display(summary_df)

print("\n✅ All 7 datasets enriched with metadata (ingestion_timestamp, source_file_name) and written to Silver layer as Delta tables!")
print(f"📁 Silver path: {silver_base_path}")
```

## Silver -> Gold
1. **Data validations** on the silver tables (critical checks stop the run, warnings are reported)
2. **Joins** to build an enriched sales fact
3. **Business aggregations** (monthly KPIs, category, region, reseller, salesperson vs target)
4. **Write to gold** as Delta tables in `adventure_works_catalog.gold`

```python
from pyspark.sql import functions as F

catalog = "adventure_works_catalog"
gold_base_path = "abfss://gold@adventurework.dfs.core.windows.net/"

product            = spark.table(f"{catalog}.silver.product")
region             = spark.table(f"{catalog}.silver.region")
reseller           = spark.table(f"{catalog}.silver.reseller")
sales              = spark.table(f"{catalog}.silver.sales")
salesperson        = spark.table(f"{catalog}.silver.salesperson")
salesperson_region = spark.table(f"{catalog}.silver.salesperson_region")
targets            = spark.table(f"{catalog}.silver.targets")

results = []
def check(name, failed_rows, critical=True):
    status = "PASS" if failed_rows == 0 else ("FAIL" if critical else "WARN")
    results.append((name, int(failed_rows), status))

# 1. Every table has data
for n, df in {"product": product, "region": region, "reseller": reseller, "sales": sales,
              "salesperson": salesperson, "salesperson_region": salesperson_region, "targets": targets}.items():
    check(f"{n}: table is not empty", 0 if df.limit(1).count() else 1)

# 2. Primary keys are not null and unique
for n, df, k in [("product", product, "ProductKey"), ("region", region, "SalesTerritoryKey"),
                 ("reseller", reseller, "ResellerKey"), ("salesperson", salesperson, "EmployeeKey")]:
    check(f"{n}: {k} not null", df.filter(F.col(k).isNull()).count())
    check(f"{n}: {k} unique", df.count() - df.select(k).distinct().count())

# 3. Sales fact rules
check("sales: required fields not null",
      sales.filter(F.col("SalesOrderNumber").isNull() | F.col("OrderDate").isNull() |
                   F.col("ProductKey").isNull() | F.col("Sales").isNull() | F.col("Cost").isNull()).count())
check("sales: Quantity > 0", sales.filter(F.col("Quantity") <= 0).count())
check("sales: Sales and Cost not negative", sales.filter((F.col("Sales") < 0) | (F.col("Cost") < 0)).count())
check("sales: OrderDate not in the future", sales.filter(F.col("OrderDate") > F.current_date()).count())
check("sales: Sales = Quantity x UnitPrice (+/- 0.01)",
      sales.filter(F.abs(F.col("Sales") - F.col("Quantity") * F.col("UnitPrice")) > 0.01).count(), critical=False)

# 4. Referential integrity (orphans are kept via left joins, so these are warnings)
for n, dim, k in [("product", product, "ProductKey"), ("reseller", reseller, "ResellerKey"),
                  ("region", region, "SalesTerritoryKey"), ("salesperson", salesperson, "EmployeeKey")]:
    check(f"sales -> {n}: orphan {k}", sales.join(dim, k, "left_anti").count(), critical=False)

# 5. Targets rules
check("targets: Target and TargetMonth not null",
      targets.filter(F.col("Target").isNull() | F.col("TargetMonth").isNull()).count())
check("targets: Target > 0", targets.filter(F.col("Target") <= 0).count())
check("targets -> salesperson: orphan EmployeeID", targets.join(salesperson, "EmployeeID", "left_anti").count(), critical=False)

validation_df = spark.createDataFrame(results, "check STRING, failed_rows LONG, status STRING")
display(validation_df)

failed = [r[0] for r in results if r[2] == "FAIL"]
if failed:
    raise ValueError(f"Critical data validation checks failed: {failed}")
print(f"Validation complete: {sum(r[2]=='PASS' for r in results)} passed, "
      f"{sum(r[2]=='WARN' for r in results)} warnings, 0 critical failures")
```

```python
fact_sales = (
    sales.select("SalesOrderNumber", "OrderDate", "ProductKey", "ResellerKey", "EmployeeKey",
                 "SalesTerritoryKey", "Quantity", "UnitPrice", "Sales", "Cost")
    .join(product.select("ProductKey", "ProductName", "Category", "Subcategory", "Color", "StandardCost"),
          "ProductKey", "left")
    .join(reseller.select("ResellerKey", "ResellerName", "BusinessType", "City", "StateProvince", "CountryRegion"),
          "ResellerKey", "left")
    .join(region.select("SalesTerritoryKey", "Region", "Country", F.col("Group").alias("RegionGroup")),
          "SalesTerritoryKey", "left")
    .join(salesperson.select("EmployeeKey", "EmployeeID", "SalespersonName", "Title"),
          "EmployeeKey", "left")
    .withColumn("Profit", F.round(F.col("Sales") - F.col("Cost"), 2))
    .withColumn("OrderYear", F.year("OrderDate"))
    .withColumn("OrderMonth", F.trunc("OrderDate", "month"))
)

# Joins must not add or drop sales rows
sales_rows, fact_rows = sales.count(), fact_sales.count()
print(f"silver sales rows: {sales_rows} | enriched fact rows: {fact_rows}")
assert sales_rows == fact_rows, "Join changed the number of sales rows (duplicate dimension keys?)"
display(fact_sales.limit(5))
```

```python
from pyspark.sql import Window

def kpis():
    """Standard revenue KPIs used by every gold aggregate."""
    return [
        F.round(F.sum("Sales"), 2).alias("TotalRevenue"),
        F.round(F.sum("Cost"), 2).alias("TotalCost"),
        F.round(F.sum("Profit"), 2).alias("TotalProfit"),
        F.countDistinct("SalesOrderNumber").alias("TotalOrders"),
        F.sum("Quantity").alias("TotalQuantity"),
    ]

def add_ratios(df):
    return (df
        .withColumn("ProfitMarginPct", F.round(F.col("TotalProfit") * 100 / F.col("TotalRevenue"), 2))
        .withColumn("AvgOrderValue", F.round(F.col("TotalRevenue") / F.col("TotalOrders"), 2)))

# 1. Monthly sales KPIs with month-over-month growth
monthly_sales_summary = add_ratios(fact_sales.groupBy("OrderYear", "OrderMonth").agg(*kpis()))
w_month = Window.orderBy("OrderMonth")
monthly_sales_summary = (monthly_sales_summary
    .withColumn("PrevMonthRevenue", F.lag("TotalRevenue").over(w_month))
    .withColumn("MoMGrowthPct", F.round((F.col("TotalRevenue") - F.col("PrevMonthRevenue")) * 100 / F.col("PrevMonthRevenue"), 2))
    .orderBy("OrderMonth"))

# 2. Product category / subcategory performance
sales_by_category = add_ratios(fact_sales.groupBy("Category", "Subcategory").agg(*kpis())).orderBy(F.desc("TotalRevenue"))

# 3. Regional performance
sales_by_region = add_ratios(fact_sales.groupBy("RegionGroup", "Country", "Region").agg(*kpis())).orderBy(F.desc("TotalRevenue"))

# 4. Reseller performance with revenue rank
reseller_performance = (add_ratios(fact_sales
        .groupBy("ResellerKey", "ResellerName", "BusinessType", "CountryRegion").agg(*kpis()))
    .withColumn("RevenueRank", F.dense_rank().over(Window.orderBy(F.desc("TotalRevenue"))))
    .orderBy("RevenueRank"))

# 5. Salesperson actual vs monthly target
actual_by_rep_month = (fact_sales
    .groupBy("EmployeeID", "OrderMonth")
    .agg(F.round(F.sum("Sales"), 2).alias("ActualRevenue")))
salesperson_target_vs_actual = (targets.select("EmployeeID", "TargetMonth", "Target")
    .join(salesperson.select("EmployeeID", "SalespersonName", "Title"), "EmployeeID", "left")
    .join(actual_by_rep_month,
          (targets.EmployeeID == actual_by_rep_month.EmployeeID) & (targets.TargetMonth == actual_by_rep_month.OrderMonth), "left")
    .select(targets.EmployeeID, "SalespersonName", "Title", "TargetMonth", "Target",
            F.coalesce("ActualRevenue", F.lit(0.0)).alias("ActualRevenue"))
    .withColumn("AttainmentPct", F.round(F.col("ActualRevenue") * 100 / F.col("Target"), 2))
    .withColumn("TargetMet", F.col("ActualRevenue") >= F.col("Target"))
    .orderBy("TargetMonth", "SalespersonName"))

display(monthly_sales_summary)
display(sales_by_category)
display(sales_by_region)
display(reseller_performance.limit(10))
display(salesperson_target_vs_actual.limit(10))
```

```python
gold_tables = {
    "fact_sales": fact_sales,
    "monthly_sales_summary": monthly_sales_summary,
    "sales_by_category": sales_by_category,
    "sales_by_region": sales_by_region,
    "reseller_performance": reseller_performance,
    "salesperson_target_vs_actual": salesperson_target_vs_actual,
}

for name, df in gold_tables.items():
    (df.withColumn("gold_load_timestamp", F.current_timestamp())
       .write.format("delta")
       .mode("overwrite")
       .option("overwriteSchema", "true")
       .option("path", f"{gold_base_path}{name}")       # external table in the gold container
       .saveAsTable(f"{catalog}.gold.{name}"))
    print(f"✓ {catalog}.gold.{name}: {spark.table(f'{catalog}.gold.{name}').count()} rows")

display(spark.sql(f"SHOW TABLES IN {catalog}.gold"))
```
